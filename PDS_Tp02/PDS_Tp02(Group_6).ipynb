```python
grade={
    "Dara":[80, 75, 90],
    "Sophea":[95, 88, 92],
    "Vibol":[40, 55, 45],
    "Malis":[70, 85, 78],
    
}
added=grade.update({"Kosal": [65, 70, 60]})
```

```python
average={}
for name,score in grade.items():
    average[name]=round(sum(score)/len(score),1)
    
print("Average : ",average)
```

```python
passed=set()

for name,avg in average.items():
    if avg>=50:
        print(name,"Passed.")
        passed.add(name)
    else:
        print(name,"Fail.")
        
print("Passed : ",sorted(passed))
```
